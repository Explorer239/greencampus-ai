# GreenCampus AI – Data Cleaning and Validation

## 1. Introduction

This notebook prepares the simulated campus sustainability dataset
for analysis and AI modelling.

The dataset is checked for missing values, duplicate records,
invalid values, and data consistency.

## 2. Load the Dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("campus_sustainability_data.csv")
print("Dataset loaded successfully!")
print("Rows:" , df.shape[0])
print("Columns:" , df.columns)

df.head()
```

## 3. Check Dataset Structure

#### Checking the column names

```python
df.columns
```

#### Checking data types

```python
df.dtypes
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.dtypes
```

## 4. Missing Value Check

```python
df.isnull().sum()
```

## 5. Duplicate Record Check

```python
df.duplicated().sum()
```

#### Checking numerical values

```python
df[["Occupancy", "Electricity_kWh", "Water_Litres", "Waste_kg", "Temperature_C"]].describe()
```

## 6. Negative Value Check

```python
print("Negative occupancy:", (df["Occupancy"] < 0).sum())

print("Negative electricity:", (df["Electricity_kWh"] < 0).sum())

print("Negative water:", (df["Water_Litres"] < 0).sum())

print("Negative Waste:", (df["Waste_kg"] < 0).sum())
```

## 7. Data Consistency Checks

#### Checking building categories

```python
df["Building"].unique()
```

```python
df["Building_Type"].unique()
```

#### Checking working day values

```python
df["Working_Day"].value_counts()
```

```python
df["Event_Day"].value_counts()
```

#### Checking date range

```python
print("Starting date:", df["Date"].min())

print("Ending date:", df["Date"].max())
```

#### Checking whether created intentional anomalies exist

```python
df[(df["Date"] == "2026-02-15") & (df["Building"] == "Block A")]
```

```python
df[(df["Date"] == "2026-04-10") & (df["Building"] == "Block B")]
```

```python
df[(df["Date"] == "2026-05-20") & (df["Building"] == "Library")]
```

#### Checking overall dataset quality

```python
print("Dataset shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate values:", df.duplicated().sum())
print("Negative values:", (df.select_dtypes(include=np.number)< 0).sum().sum())
```

## 8. Final Cleaned Dataset

After validation, the cleaned dataset is saved for further analysis.

```python
df.to_csv("cleaned_campus_sustainability_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
